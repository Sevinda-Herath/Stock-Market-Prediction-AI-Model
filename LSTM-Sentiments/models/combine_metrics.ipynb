```python
import pandas as pd
import os

# List of stock symbols
top_10_symbols = [
    "AMZN",        # Amazon
    "AAPL",        # Apple
    "GOOGL",       # Alphabet (Class A)
    "005930.KS",   # Samsung Electronics (Korea Exchange)
    "2317.TW",     # Foxconn (Hon Hai Precision, Taiwan)
    "MSFT",        # Microsoft
    "JD",          # Jingdong (JD.com)
    "BABA",        # Alibaba (NYSE)
    "T",           # AT&T
    "META"         # Meta (Facebook)
]

# Empty list to collect DataFrames
all_data = []

# Loop through symbols and load corresponding CSVs
for symbol in top_10_symbols:
    filename = f"metrics/{symbol}_lstm_model_metrics.csv"
    if os.path.exists(filename):
        df = pd.read_csv(filename)
        all_data.append(df)
    else:
        print(f"File not found: {filename}")

# Combine all rows into a single DataFrame
combined_df = pd.concat(all_data, ignore_index=True)

# Save the combined data to a new CSV file
combined_df.to_csv(f"metrics/combined_lstm_model_metrics.csv", index=False)

print("✅ Combined file saved as 'combined_lstm_model_metrics.csv'")
```

```python
import pandas as pd

# Read the combined metrics file
data = pd.read_csv("metrics/combined_lstm_model_metrics.csv")

# Rearranged column order
reordered_columns = [
    "symbol", "train_mae", "test_mae",
    "train_mse", "test_mse",
    "train_r2", "test_r2",
    "next_day_prediction"
]

# Apply the new column order
df = data[reordered_columns]


# Display styled table
df.style.set_caption("📈 LSTM Model Metrics Comparison") \
    .format(precision=4) \
    .background_gradient(subset=["train_r2", "test_r2"], cmap="Greens") \
    .background_gradient(subset=["test_mae", "test_mse"], cmap="Reds_r") \
    .bar(subset=["next_day_prediction"], color="#3498db")
```
